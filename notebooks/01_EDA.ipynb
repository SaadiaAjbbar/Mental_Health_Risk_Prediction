Vérifier les valeurs manquantes, les doublons, les incohérences et les valeurs aberrantes.

```python
missing_values = df.isnull().sum()

print("Nombre de valeurs manquantes par colonne :")
display(missing_values)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Pourcentage de valeurs manquantes par colonne :")
display(missing_percentage.round(2))
```

```python
missing_columns = missing_percentage[missing_percentage > 0]

print("Colonnes contenant des valeurs manquantes :")
display(missing_columns)
```

```python
duplicate_count = df.duplicated().sum()

print("Nombre de lignes dupliquées :", duplicate_count)
```

```python
categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns

for column in categorical_columns:
    print(f"\nColonne : {column}")
    print(df[column].value_counts(dropna=False))
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    print(f"\nColonne : {column}")
    print("Minimum :", df[column].min())
    print("Maximum :", df[column].max())
```

```python
screen_column = "Screen_Time_Hours"
social_column = "Social_Media_Hours"

incoherent_screen = df[screen_column] < df[social_column]

incoherent_count = incoherent_screen.sum()

print("Nombre de cas incohérents :", incoherent_count)
print("Pourcentage de cas incohérents :",
      round((incoherent_count / len(df)) * 100, 2), "%")
```

```python
if incoherent_count == 0:

    print("Aucune incohérence détectée.")

elif incoherent_count < 10:

    df = df.loc[~incoherent_screen].copy()

    print(incoherent_count, "lignes incohérentes supprimées.")

else:

    df.loc[incoherent_screen, screen_column] = df.loc[
        incoherent_screen,
        [screen_column, social_column]
    ].max(axis=1)

    print(incoherent_count, "valeurs de Screen_Time_Hours corrigées.")
```

```python
weekly_screen_hours = df["Screen_Time_Hours"] * 7

total_weekly_hours = (
    weekly_screen_hours + df["Study_Hours_Per_Week"]
)

incoherent_weekly_time = total_weekly_hours > 168

print("Nombre de cas dépassant 168 heures :",
      incoherent_weekly_time.sum())

print("Pourcentage de cas concernés :",
      round(incoherent_weekly_time.mean() * 100, 2), "%")
```

```python
outliers_summary = {}

for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    outliers_summary[column] = len(outliers)

print("Nombre d'outliers potentiels par variable :")

for column, count in outliers_summary.items():
    print(f"{column} : {count}")
```

Examiner les corrélations entre les variables.

```python
correlation_matrix = X[numeric_features].corr(
    method="pearson"
)

display(correlation_matrix)
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Matrice de corrélation entre les variables")
plt.tight_layout()
plt.show()
```

```python
correlation_pairs = correlation_matrix.unstack()

correlation_pairs = correlation_pairs[
    (correlation_pairs.abs() >= 0.7) &
    (correlation_pairs.abs() < 1)
]

display(
    correlation_pairs.sort_values(
        key=abs,
        ascending=False
    )
)
```

identifier les variables les plus fortement liées au niveau de risque ;

```python
risk_mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}

df_analysis = df.copy()

df_analysis[TARGET_COLUMN] = df_analysis[TARGET_COLUMN].map(
    risk_mapping
)
```

```python
print("Valeurs manquantes après mapping :",
      df_analysis[TARGET_COLUMN].isnull().sum())
```

```python
spearman_correlations = df_analysis.corr(
    method="spearman",
    numeric_only=True
)

target_correlations = spearman_correlations[TARGET_COLUMN]

target_correlations = target_correlations.drop(
    TARGET_COLUMN
)

target_correlations = target_correlations.sort_values(
    key=abs,
    ascending=False
)

display(target_correlations)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=target_correlations.values,
    y=target_correlations.index
)

plt.title("Corrélation des variables avec le niveau de risque")
plt.xlabel("Coefficient de corrélation de Spearman")
plt.ylabel("Variables explicatives")

plt.axvline(x=0, color="black", linewidth=1)

plt.tight_layout()
plt.show()
```

```python
top_5_variables = target_correlations.abs().sort_values(
    ascending=False
).head(5)

print("Les 5 variables les plus fortement liées au niveau de risque :")

display(
    target_correlations.loc[top_5_variables.index]
)
```

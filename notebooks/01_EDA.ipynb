```python
outliers_summary = {}

for column in numeric_columns:
    
    Q1 = df[column].quantile(0.25)
    
    Q3 = df[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_limit = Q1 - 3 * IQR
    upper_limit = Q1 + 3 * IQR
    

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    outliers_summary[column] = len(outliers)

print("Nombre d'outliers potentiels par colonne :")

for column, count in outliers_summary.items():

    print(f"{column} : {count}")
```

```python
correlation_matrix = df[numeric_columns].corr(method="pearson")

display(correlation_matrix)
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Matrice de corrélation entre les variables")
plt.tight_layout()
plt.show()
```

```python
correlation_pairs = correlation_matrix.unstack()

correlation_pairs = correlation_pairs[
    (correlation_pairs.abs() >= 0.7) &
    (correlation_pairs.abs() < 1)
]

display(correlation_pairs.sort_values(
    key=abs,
    ascending=False
))
```

```python
print("Type de la cible :", df[TARGET_COLUMN].dtype)

print("\nValeurs uniques :")
print(df[TARGET_COLUMN].unique())

print("\nNombre de classes :", df[TARGET_COLUMN].nunique())
```

```python
risk_mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}
```

```python
df_analysis = df.copy()

df_analysis[TARGET_COLUMN] = df_analysis[TARGET_COLUMN].map(
    risk_mapping
)

spearman_correlations = df_analysis.corr(
    method="spearman",
    numeric_only=True
)

target_correlations = spearman_correlations[TARGET_COLUMN]

target_correlations = target_correlations.drop(TARGET_COLUMN)

target_correlations = target_correlations.sort_values(
    key=abs,
    ascending=False
)

display(target_correlations)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=target_correlations.values,
    y=target_correlations.index
)

plt.title("Corrélation des variables avec le niveau de risque")
plt.xlabel("Coefficient de corrélation de Spearman")
plt.ylabel("Variables explicatives")
plt.axvline(x=0, color="black", linewidth=1)

plt.tight_layout()
plt.show()
```

```python
top_5_variables = target_correlations.abs().sort_values(
    ascending=False
).head(5)

print("Les 5 variables les plus fortement liées au niveau de risque :")

display(
    target_correlations.loc[top_5_variables.index]
)
```
